# Feature Engineering

```python
#Total crimes per row
crime_cols = ['Rape','K&A','DD','AoW','AoM','DV','WT']
df['Total_crime'] = df[crime_cols].sum(axis=1)
```

```python
# crime share(%)
for col in crime_cols:
    df[col+'_pct'] =(df[col]/df['Total_crime'])*100
```

# 12 CAGR Comparison: DV vs Rape Vs K&A

```python
crime_cols = ['DV','Rape','K&A']
yearly_crime = (df.groupby('Year')[crime_cols].sum().sort_index())

def calculate_cagr(series):
    start = series.iloc[0]
    end = series.iloc[-1]
    years = len(series)-1
    return ((end/start)**(1/years)-1)*100
```

```python
cagr_results = {}
for crime in crime_cols:
    cagr_results[crime] = calculate_cagr(yearly_crime[crime])

cagr_df =(pd.DataFrame.from_dict(cagr_results,orient ='index',columns=['CAGR (%)']).round(2))
cagr_df
```

```python
cagr_df.plot(kind='bar',figsize=(6,4))
plt.title("CAGR Comparison of Major Crime Types (2001-2021)")
plt.ylabel("CAGR (%)")
plt.xlable("Crime Type")
plt.show()
```
